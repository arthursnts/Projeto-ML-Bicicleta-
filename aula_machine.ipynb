# Dar um exemplo de algoritmo de Scikit-learn, demonstrar um exemplo de gráfico.

```python
from sklearn.linear_model import Lasso


x = [[10_000], [15_000], [30_000], [42_000], [47_000], [67_000]]
#Variação 
y = [74_000, 80_000, 120_000, 150_000, 147_000, 210_000]

#Criação do modelo
modelo = Lasso()
modelo.fit(x,y)

verba_marketing_previsao = [[1200]]
lucro_previsto = modelo.predict(verba_marketing_previsao)

print(f"Investindo R$ {verba_marketing_previsao[0][0]:,.2f} Teremos lucro de R$ {lucro_previsto[0]:,.2f} ")
```

```python
from sklearn.linear_model import LinearRegression

duracoes = [10, 15, 20, 25, 30, 35, 40, 50, 55, 60]
precos = [10.00, 13.00, 14.00, 18.00, 19.00, 23.00, 24.00, 30.00, 32.00, 34.00]

X = [[duracao] for duracao in duracoes]
y = precos

modelo = LinearRegression()
modelo.fit(X, y)

previsao_45 = modelo.predict([[45]])
print(f"Preço estimado para 45 minutos: R$ {previsao_45[0]:.2f}")

previsao_20 = modelo.predict([[20]])
print(f"Previsão para 20 minutos: R$ {previsao_20[0]:.2f}")
print("Preço histórico para 20 minutos: R$ 14,00")

previsao_60 = modelo.predict([[60]])
print(f"Previsão para 60 minutos: R$ {previsao_60[0]:.2f}")
print("Preço histórico para 60 minutos: R$ 34,00")
```
